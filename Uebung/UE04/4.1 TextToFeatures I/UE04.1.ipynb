```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp("Jim loves NLP. He will learn NLP in two monts. NLP is future.")

import sklearn
from sklearn.feature_extraction.text import CountVectorizer
text="Jim loves NLP. He will learn NLP in two monts. NLP is future." 
 
def get_spacy_tokens(text): 
    doc = nlp(text) 
    return [token.text for token in doc] 
 
text_tokens = get_spacy_tokens(text) 
 
vectorizer = CountVectorizer(tokenizer=get_spacy_tokens, lowercase=False) 
 
# Erstellung des Vokabulars: 
vectorizer.fit(text_tokens) 
print("Vocabulary: ", vectorizer.vocabulary_) 
 
# Erstellung der Matrix mit Wortvektoren (One Hote Encoding): 
vector = vectorizer.transform(text_tokens)   
print("Encoded Document is:") 
print(vector.toarray())
```

```python
# 4.1
import keras
import tensorflow
from tensorflow.keras.utils import to_categorical

def one_hot_encoding(sentence, vocabulary):
    words = sentence.split()
    integer_encoded = [vocabulary.index(word) for word in words]
    one_hot_encoded = to_categorical(integer_encoded, num_classes=len(vocabulary))
    return one_hot_encoded

one_hot_encoding("Jim will learn NLP in future.", ["Jim", "NLP", "future", "learn", "in", "will"])
```
