```python
with open('schachnovelle.pdf', 'rb') as pdf:
    reader = pypdf.PdfFileReader(pdf)
    print(f"Author:     {getattr(reader.documentInfo, 'author')}\nProducer:   {getattr(reader.documentInfo, 'producer')}\nTitle:      {getattr(reader.documentInfo, 'title')}")
```
