```python
# 3.3.2

import spacy
from spacy.matcher import Matcher
nlp = spacy.load("en_core_web_sm")
matcher = Matcher(nlp.vocab)
text = open("dataScientistJobDesc.txt", encoding="utf8").read()
pattern1 = [{"LEMMA":"experience"}, {"IS_STOP":True, "OP":"*"}, {"POS":"PROPN", "OP":"+"}]
pattern2 = [{"LEMMA":"experience"}, {"IS_STOP":True, "OP":"*"}, {"POS":"NOUN", "OP":"+"}]
pattern3 = [{"POS":"PROPN", "OP":"+"}, {"LEMMA":"experience"}]
pattern4 = [{"POS":"NOUN", "OP":"+"}, {"LEMMA":"experience"}]
matcher.add("PATTERN1", [pattern1])
matcher.add("PATTERN2", [pattern2])
matcher.add("PATTERN3", [pattern3])
matcher.add("PATTERN4", [pattern4])
doc = nlp(text)
matches = matcher(doc)
for match_id, start, end in matches:
    string_id = nlp.vocab.strings[match_id]
    span = doc[start:end]
    print(match_id, string_id, start, end, span.text)
```
