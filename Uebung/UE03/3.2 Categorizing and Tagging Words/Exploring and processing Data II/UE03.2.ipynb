```python
# 3.2.1

import spacy
from spacy.lang.en import STOP_WORDS
nlp = spacy.load("en_core_web_sm")

import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.tag import pos_tag
nltk.download('averaged_perceptron_tagger')
nltk.download('stopwords')

with open ("dataScientistJobDesc.txt", "r") as txt:
    data = txt.read()
    doc = nlp(data)

    # spacy
    spacy_extracted_words = []
    for token in doc:
        if not (token.is_stop or token.is_punct):
            if token.pos_ == "PROPN" or token.pos_ == "NOUN":
                spacy_extracted_words.append(token.text)

    # nltk
    tokens = word_tokenize(data)
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token.lower() not in stop_words]
    tagged_tokens = pos_tag(tokens)
    nltk_extracted_words = []
    for token, tag in tagged_tokens:
        if tag == "NN" or tag == "NNP":
            nltk_extracted_words.append(token)

    print(f"Spacy: {len(spacy_extracted_words)}, NLTK: {len(nltk_extracted_words)}")
```

```python
# 3.2.2

import spacy
nlp = spacy.load("en_core_web_sm")

with open ("dataScientistJobDesc.txt", "r") as txt:
    data = txt.read()
    doc = nlp(data)

    pos_counts = {}

    for token in doc:
        if not (token.is_stop or token.is_punct):
            pos = token.pos_
            if pos in pos_counts:
                pos_counts[pos] += 1
            else:
                pos_counts[pos] = 1

    print(pos_counts)
```

```python
# 3.2.3

import spacy
from collections import Counter
nlp = spacy.load("en_core_web_sm")

with open ("dataScientistJobDesc.txt", "r") as txt:
    data = txt.read()
    doc = nlp(data)
    
    word_counter = Counter()

    for token in doc:
        if not (token.is_stop or token.is_punct):
            word_counter[token.text.lower()] += 1

    most_common_word = word_counter.most_common(1)[0][0]

    print(f"Most common word: '{most_common_word}'")
```
