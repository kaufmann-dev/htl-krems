```python
# 3.9 #

tweets=[ 
    'This is introduction to NLP', 
    'It is likely to be useful, to people ', 
    'Machine learning is the new electrcity', 
    'There would be less hype around AI and more action going forward', 
    'python is the best tool!','R is good langauage', 
    'I like this book','I want more books like this' 
]
tweets_new = []
for tweet in tweets:
    s = ""
    bruh = [word for word in tweet.split() if word not in stop]
    for word in bruh:
        s += word + " "
    tweets_new.append(s)
 
tweetsdf = pd.DataFrame()
tweetsdf['tweets'] = tweets_new
tweetsdf['tweets'] = tweetsdf['tweets'].str.lower().replace('[.,!?;]','', regex=True)
display(tweetsdf.head(7))
```

```python
# 3.10 #

from nltk.stem import PorterStemmer as pt
stemmer = pt()
 
text=['I like fishing','I eat fish','There are many fishes in pound'] 
for i in text:
    gru = [stemmer.stem(j) for j in i.split()]
    s = ""
    for minion in gru:
        s += minion + " "
    print(s)
```

```python
# 3.11 #

from nltk.stem import WordNetLemmatizer 
lemmatizer = WordNetLemmatizer()
text=['I like fishing','I eat fish','There are many fishes in pound'] 
for i in text:
    gru = [lemmatizer.lemmatize(j) for j in i.split()]
    s = ""
    for minion in gru:
        s += minion + " "
    print(s)
```

```python
# Text Preprocessing Pipeline #

import re
def preprocess_text(text_to_process, processing_function_list):
    for function in processing_function_list:
        match function:
            case "to_lower":
                text_to_process = text_to_process.lower()
            case "remove_url":
                text_to_process = re.sub(r"http\S+", "", text_to_process)
            case "remove_email":
                text_to_process = re.sub(r"\S+@\S+", "", text_to_process)
            case "remove_punctuation":
                text_to_process = re.sub(r"[.,!?;]", "", text_to_process)
            case "remove_stopword":
                text_to_process = " ".join([word for word in text_to_process.split() if word not in stop])
            case "lemmatize_word":
                text_to_process = " ".join([lemmatizer.lemmatize(word) for word in text_to_process.split()])
    return text_to_process

preprocess_functions = ["to_lower", "remove_url", "remove_email", "remove_punctuation"]
preprocessed_text = preprocess_text("Lorem ipsum dolor sit amet, consetetur sadipscing elitr, sed diam nonumy eirmod tempor invidunt ut labore et dolore magna aliquyam erat, sed diam voluptua.", preprocess_functions)
print(preprocessed_text)
```
