```python
# 3.3.2

# GPE: „Geo-Political Entity“. Wird verwendet, um Einheiten zu kennzeichnen, die einen bestimmten geografischen Ort darstellen, wie Länder, Städte und Regionen.

# NORP: „Nationalities or Religious or Political Groups“. Wird verwendet, um Einheiten zu kennzeichnen, die eine Nationalität, religiöse oder politische Gruppe repräsentieren, wie z. B. „Muslime“ oder „Republikaner“.

# ORDINAL: Wird verwendet, um Entitäten zu kennzeichnen, die eine Ordnungszahl darstellen, z. B. „erste“ oder „zweite“. Es kann verwendet werden, um die in einem bestimmten Text verwendeten Ordnungszahlen zu identifizieren.

import spacy
from spacy.matcher import Matcher
nlp = spacy.load("en_core_web_sm")
matcher = Matcher(nlp.vocab)
text = open("dataScientistJobDesc.txt", encoding="utf8").read()
pattern1 = [{"LEMMA":"experience"}, {"IS_STOP":True, "OP":"*"}, {"POS":"PROPN", "OP":"+"}]
pattern2 = [{"LEMMA":"experience"}, {"IS_STOP":True, "OP":"*"}, {"POS":"NOUN", "OP":"+"}]
pattern3 = [{"POS":"PROPN", "OP":"+"}, {"LEMMA":"experience"}]
pattern4 = [{"POS":"NOUN", "OP":"+"}, {"LEMMA":"experience"}]
matcher.add("PATTERN1", [pattern1])
matcher.add("PATTERN2", [pattern2])
matcher.add("PATTERN3", [pattern3])
matcher.add("PATTERN4", [pattern4])
doc = nlp(text)
matches = matcher(doc)
for match_id, start, end in matches:
    string_id = nlp.vocab.strings[match_id]
    span = doc[start:end]
    print(match_id, string_id, start, end, span.text)
```

```python
# 3.3.3
import spacy

nlp = spacy.load("en_core_web_md")

text = "Dear Joe! I have organized a meeting with Elon Musk from Siemens for tomorrow. Meeting place is Vienna."
doc = nlp(text)

for entity in doc.ents:
    if entity.label_ == "PERSON":
        print(entity.text)
```

```python
# Zugriff  auf  Entitäts-Informationen  ist  über  die  Token-Ebene
for token in doc:     
    print(f'{token.text:10} {token.ent_iob_} {token.ent_type_}')
```

```python
# 3.3.4
import spacy

nlp = spacy.load("en_core_web_md")

text = "Der Ministerrat der Republik Österreich beschloss am 25.März 2014, eine „Unabhängige Untersuchungskommission zur transparenten Aufklärung der Vorkommnisse rund um die Hypo Group Alpe-Adria“ einzusetzen. Die Untersuchungskommission (Manuel Ammann, Carl Baudenbacher, Ernst Wilhelm Contzen, Irmgard Griss, Claus-Peter Weber) hat, beginnend mit 1.Mai 2014, durch Auswertung von beigeschafften Unterlagen und allgemein zugänglichen Quellen sowie durch Befragung von Auskunftspersonen den maßgeblichen Sachverhalt festgestellt und nach fachlichen Kriterien bewertet."
doc = nlp(text)

for entity in doc.ents:
    if entity.label_ == "PERSON":
        text = text.replace(entity.text, "<REDACTED>")
        
print(text)
```
