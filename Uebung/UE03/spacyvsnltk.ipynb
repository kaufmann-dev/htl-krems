```python
# NLTK #

import nltk
text = "The quick brown fox jumps over the lazy dog."


# Regex #

from nltk.tokenize import regexp_tokenize
print(regexp_tokenize(text, pattern='\w+'))

from nltk.tokenize import RegexpTokenizer
tokenizer = RegexpTokenizer(pattern='\w+')
print(tokenizer.tokenize(text))

# Split text into tokens #

print(text.split())

from nltk.tokenize import WhitespaceTokenizer
tokenizer = WhitespaceTokenizer()
print(tokenizer.tokenize(text))

# Remove punctuation / only words #

from nltk.tokenize import word_tokenize
print(word_tokenize(text))
from nltk.tokenize import WordPunctTokenizer
tokenizer = WordPunctTokenizer()
print(tokenizer.tokenize(text))

# Only sentences #

from nltk.tokenize import sent_tokenize
print(sent_tokenize(text))

from nltk.tokenize import PunktSentenceTokenizer
tokenizer = PunktSentenceTokenizer()
print(tokenizer.tokenize(text))

# Stopwords #

nltk.download('stopwords') 
from nltk.corpus import stopwords
stop = stopwords.words('english')  
print([word for word in text.split() if word not in stop] )

# Stemming #

from nltk.stem import PorterStemmer as pt
stemmer = pt()
print([stemmer.stem(word) for word in text.split()]) 

# Lemmatiztion #

nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
print([lemmatizer.lemmatize(word) for word in text.split()])

# Part of Speech (POS) Tagging #
print(nltk.pos_tag(text.split()))
```

```python
# Spacy

import spacy
text = "The quick brown fox jumps over the lazy dog."
nlp = spacy.load('en_core_web_sm')
doc=nlp(text)

# Remove punctuation / only words #

print([token.text for token in doc if not token.is_punct])

# Stopwords #

print([token.text for token in doc if not token.is_stop])

# Lemmatization #

print([token.lemma_ for token in doc])

# POS Tagging # 

print([(token.text, token.pos_) for token in doc])

# Pettern matching #

from spacy.matcher import Matcher 
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "ADJ"}, {"POS": "ADJ", "OP": "?"}, {"POS": "NOUN"}]
matcher.add("pattern_1", [pattern])
print([doc[start:end] for match_id, start, end in matcher(doc)])

# Match all adjectives #
{"POS": "ADJ"}
# Match zero or one adjectives #
{"POS": "ADJ", "OP": "?"}
# Match one or more adjectives #
{"POS": "ADJ", "OP": "+"}
# Match zero or more adjectives #
{"POS": "ADJ", "OP": "*"}
# Match one or more adjectives, but not more than 3 #
{"POS": "ADJ", "OP": "+{1,3}"}
# Match zero or more adjectives, but not more than 3 #
{"POS": "ADJ", "OP": "*{,3}"}
# Match lemma of "love" #
{"LEMMA": "love"}
# Match lemma of "love" or "like" #
{"LEMMA": {"IN": ["love", "like"]}}
# Match lemma of "love" or "like", but not "loved" or "liked" #
{"LEMMA": {"IN": ["love", "like"], "NOT_IN": ["loved", "liked"]}}
# Match any token with the lowercase form "hello" #
{"LOWER": "hello"}
# Match text #
{"TEXT": "hello"}
# Match regular expression #
{"TEXT": {"REGEX": "hello.*"}}
# Match dictionary of token attributes #
{"ORTH": "hello", "POS": "INTJ"}
# Match digit #
{"IS_DIGIT": True}
# Match punctuation #
{"IS_PUNCT": True}
# Match whitespace #
{"IS_SPACE": True}
# Match any token that is not a digit, punctuation or whitespace #
{"IS_ALPHA": True}
# Match any ascii token #
{"IS_ASCII": True}
# Match any token that is not a stop word #
{"IS_STOP": False}

```
