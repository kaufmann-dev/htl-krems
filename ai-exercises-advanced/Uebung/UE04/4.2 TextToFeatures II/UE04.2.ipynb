```python
# 4.2.2 mit nltk

import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import Counter

# Laden der CSV-Datei in ein Pandas DataFrame
df = pd.read_csv('realdonaldtrump.csv')

# Erstellen einer Liste von Stopwörtern
stop_words = set(stopwords.words('english'))

# Erstellen eines Lemmatizers
lemmatizer = WordNetLemmatizer()

# Erstellen einer Liste von Token aus dem Text + Entfernen von Stopwörtern
tokens = [word.lower() for word in nltk.word_tokenize(df['content'].str.cat()) if word.lower() not in stop_words]

# Lemmatisierung der Token
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens]

# Ausgabe der 10 häufigsten Bigramme
bigrams = Counter(nltk.bigrams(lemmatized_tokens)).most_common(10)
print("10 häufigsten Bigramme: ", bigrams)

# Ausgabe der 10 häufigsten Trigramme
trigrams = Counter(nltk.trigrams(lemmatized_tokens)).most_common(10)
print("10 häufigsten Trigramme: ", trigrams)
```

```python
# 4.2.2 mit spacy

import spacy
import pandas as pd
from collections import Counter

# Erstellen einer Liste von Stopwörtern
nlp = spacy.load('en_core_web_sm')
stop_words = spacy.lang.en.stop_words.STOP_WORDS

# Laden der CSV-Datei in ein Pandas DataFrame
df = pd.read_csv('realdonaldtrump.csv')

# Preprocessing
texts = df['content'].tolist()
processed_texts = []
for text in texts:
    doc = nlp(text.lower())
    processed_text = []
    for token in doc:
        if not token.is_stop and token.is_alpha:
            processed_text.append(token.lemma_)
    processed_texts.append(processed_text)

# Bi- und Trigramme zählen
bi_trigrams = Counter()
for text in processed_texts:
    for i in range(len(text) - 2):
        bi_trigrams[(text[i], text[i+1])] += 1
    for i in range(len(text) - 3):
        bi_trigrams[(text[i], text[i+1], text[i+2])] += 1

# Ausgabe der häufigsten Bi- und Trigramme
print(bi_trigrams.most_common(10))
```
